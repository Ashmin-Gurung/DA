```python
# 1. Create a 3×3 matrix with values from 1 to 9
arr1 = np.arange(1, 10).reshape(3, 3)
print(arr1)
```

```python
# 2. Create an array of 10 random floats (0 to 1) and round to 2 decimals
arr = np.random.rand(10)
arr = np.round(arr, 2)
arr
```

```python
# 3. Find all even numbers using Boolean indexing
arr3 = np.array([1, 3, 6, 9, 12, 15])
even_numbers = arr3[arr3 % 2 == 0]
print(even_numbers)
```

```python
# 4. Create a 4x4 matrix filled with 7
matrix = np.full(shape=(4,4), fill_value=7)
matrix
```

```python
# 5. 5.Create a 3D array of shape (2, 3, 4) filled with random integers between 1 and 100
arr5 = np.random.randint(1, 101, size=(2, 3, 4))
print(arr5)
```

```python
# 6. extract all elements greater than 50 and replace them with -1
arr = np.random.randint(10, 100, size=(5, 5))
print(arr)
arr[arr > 50] = -1
print(arr)
```

```python
# 7. Create a 3D array of shape (3, 4, 5) filled with random integers between 1 and 100.
arr = np.random.randint(1, 101, size=(3, 4, 5))
print(arr)
```

```python
# Replace all elements greater than the mean with 0
mean_val = arr.mean()
arr[arr > mean_val] = 0
print("\nAfter replacing values greater than mean with 0:\n", arr)
```

```python
# Sum along axis=1
sum_axis1 = arr.sum(axis=1)
print("\nSum along axis=1:\n", sum_axis1)
```

```python
# Extract a new 2D array that contains only the first and last columns from each row of each depth slice.
result_2d = arr[:, :, [0, -1]].reshape(-1, 2) # fancyindexing
print(result_2d)
```

10 basic NumPy Questions

```python
# 1.Create a NumPy array of numbers from 10 to 50
arr1 = np.arange(10, 51)
print(arr1)
```

```python
#2. Generate a 3×3 identity matrix
identity = np.eye(3)
print(identity)
```

```python
# 3. Create a NumPy array of 10 random integers between 1 and 100
arr3 = np.random.randint(1, 101, 10)
print(arr3)
```

```python
#4. Find the maximum and minimum of an array
arr = np.array([5, 10, 3, 25, 8])
print("Max:", arr.max())
print("Min:", arr.min())
```

```python
# 5. Reshape a 1D array of 12 elements into a 3×4 matrix
arr5 = np.arange(1, 13).reshape(3, 4)
print(arr5)
```

```python
# 6. Create an array of all even numbers between 50 and 100
even_arr = np.arange(50, 101, 2)
print(even_arr)
```

```python
# 7. Square each element of [1, 2, 3, 4, 5]
arr7 = np.array([1, 2, 3, 4, 5])
squared = arr7 ** 2
print(squared)
```

```python
# 8. Create a 5×5 random array and find its mean
arr8 = np.random.rand(5, 5)
print(arr8)
print("Mean:", arr8.mean())
```

```python
# 9. Find the index of the maximum element
arr9 = np.array([10, 20, 25, 40, 30])
index = np.argmax(arr9)
print(index)
```

```python
# 10. Element-wise operations on two arrays
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```
