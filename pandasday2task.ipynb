1. Create a Pandas DataFrame from a dictionary of lists.

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Ram", "Hari", "Sita"],
    "Age": [23, np.nan, 25],
    "Department": ["IT", "HR", "IT"]
}

df = pd.DataFrame(data)
df
```
