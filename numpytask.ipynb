```python
import numpy as np
```

Creating NumPy 1-D array Using Python list and tuple

```python
# creating list of numbers
numbers = [1, 2, 5, 4, 3, 6]

# creating numpy array using numbers list
nparray = np.array(numbers)

print(nparray)
```

```python
# creating tuple of numbers
numbers = (1, 2, 5, 4, 3, 6)

# creating numpy array using numbers list
nparray = np.array(numbers)

print(nparray)
```

Creating NumPy 1-D array from scratch

```python
# creating an array with 10 elements filled with zeros
array1 = np.zeros(10) 

print(array1)
print(type(array1[0])) # type of first element in the array
```

```python
# creating an array with 10 elements filled with ones
array1 = np.ones(10) 

print(array1)
```

```python
# Using arange
array1 = np.arange(10)
print(array1)

# creating an array with values from 3 to 20 with a step of 2.3
array2 = np.arange(3, 20, 2.3)
print(array2)
```

np.linspace() function creates an array with evenly spaced numbers within a range.

```python
print(np.linspace(1,10,8))
```

```python
array1 = np.linspace(1, 120, 5)

print(array1)
```

create an array with 6 elements between 1 and 10

```python
a1 = np.linspace(1,10,6)
a1
```

Perform array slicing

```python
arr = np.array([1,2,3,4,5])
arr[2:4]
```

Create an array of number from 1  to 10 and print only even numbers from that

```python
arr = np.arange(1,11)
even_num = arr[arr % 2 == 0]

even_num
```
