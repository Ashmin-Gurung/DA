Implement 2d and 3d slicing

```python
import numpy as np
```

```python
arr = np.arange(16).reshape(4, 4)
print(arr)
```

```python
print(arr[:2, :]) # firdt 2 row
```

```python
print(arr[:, 2:]) # last two column
```

```python
print(arr[1:3, 1:3]) # middle sblock
```

```python
print(arr[-1, :]) # last row only
```

3d array:

```python
arr3d = np.arange(24).reshape(2, 3, 4)
print(arr3d)
```

```python
print(arr3d[0]) # first block
```

```python
print(arr3d[:, 0, :]) # first row from all block
```

```python
print(arr3d[:, :, 2]) # column 2 fromm all block and row
```

```python
print(arr3d[:, 1:3, 1:3])
```

```python
print(arr3d[:, ::2, ::2]) # slicing every second element
```
