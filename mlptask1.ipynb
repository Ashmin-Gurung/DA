1. How do you plot a simple line graph using Matplotlib?

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,10,20)
y = np.sin(x)

plt.plot(x,y)
plt.show()
```

2. Difference between plt.plot() and plt.scatter()?

```python
plt.plot(x, y)       # connects points with a line
```

```python
a = np.random.rand(50)
b = np.random.rand(50)
plt.scatter(a,b) # shows individual points with no lines
```

3. How do you change line color and style?

```python
plt.plot(x, y, color='red', linestyle='--' )
```

4. How can you set x-axis and y-axis labels?

```python
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
```

5. Which function is used to create histograms?

```python
data = [1,2,3,4,5,1,2,3,6,7,8,1,2,3]
plt.hist(data, edgecolor = "Red", color="black")
plt.show()
```

6. How do you add a legend to a plot?

```python
plt.plot(x, y, label='Sales')
plt.legend()
```

7. What does the alpha parameter control?

```python
plt.plot(x, y, alpha=0.5) # contrlling transparency
```

8. How do you plot multiple subplots in the same figure?

```python
plt.subplot(1, 2, 1) # subplot(rows, columns, position)
plt.plot(x, y)

plt.subplot(1, 2, 2)
plt.scatter(a, b)

plt.show()
```

9. Difference between plt.bar() and plt.barh()?

```python
x=[1,2,3,4]
y=[10,15,20,30]
plt.bar(x, y) # vertical bar chart
```

```python
plt.barh(x, y) # horizontal bar chart
```
